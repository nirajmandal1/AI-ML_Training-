```python
x=np.array([10,20,30,40,50])
y=np.array([100,200,300,400,500])
```

```python
plt.figure(dpi=300)
plt.plot(x,y)
plt.show
```

```python
plt.figure(facecolor="blue")
plt.plot(x,y,marker="o",linestyle="--",color="red")
plt.title("simple line chart")
plt.xlabel("quantity",color="green")
plt.ylabel("price")

plt.show()
```

```python
plt.figure(facecolor="yellow")
plt.plot(x,y,marker="o",linestyle="--",color="red",mfc="blue", mec="green",ms=10)
plt.title("simple line chart")
plt.xlabel("quantity",color="green")
plt.ylabel("price",color="green")

plt.show()
```

```python
plt.figure(facecolor="blue")
plt.plot(x,y,marker="o",linestyle="--",color="red")
plt.title("simple line chart",color="k",fontsize=20)
plt.xlabel("quantity",color="green",fontsize=20)
plt.ylabel("price",fontsize=20)
plt.grid( axis="x",linestyle="--",color="y")
plt.show()
```

```python
plt.figure(facecolor="blue")
plt.plot(x,y,marker="o",linestyle="--",color="red",linewidth=10)
plt.title("simple line chart",color="k",fontsize=20)
plt.xlabel("quantity",color="green",fontsize=20)
plt.ylabel("price",fontsize=20)
plt.grid( axis="x",linestyle="--",color="y")
plt.show()
```
